```python
# Instalação e download de recursos (rodar 1x)
!pip install -q nltk spacy pandas
!python -m spacy download pt_core_news_sm -q

import nltk
import spacy
import pandas as pd

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('rslp', quiet=True)

nlp = spacy.load("pt_core_news_sm")

print("Ambiente configurado com sucesso.")
```

```python
frase_ancora = "O banco aprovou o empréstimo ontem."

doc_ancora = nlp(frase_ancora)
tokens_sem_pontuacao = [token.text for token in doc_ancora if not token.is_punct]

print("Tokens sem pontuação:", tokens_sem_pontuacao)
```

```python
from nltk.corpus import stopwords

stop_pt = set(stopwords.words('portuguese'))
print(f"Total de stopwords em português (NLTK): {len(stop_pt)}")
print(list(stop_pt)[:15])
```

```python
tokens_sem_stopwords_nltk = [t for t in tokens_sem_pontuacao if t.lower() not in stop_pt]

print("Antes: ", tokens_sem_pontuacao)
print("Depois:", tokens_sem_stopwords_nltk)
```

```python
tokens_sem_stopwords_spacy = [token.text for token in doc_ancora if not token.is_stop and not token.is_punct]

print("Tokens sem stopwords/pontuação (spaCy):", tokens_sem_stopwords_spacy)
```

```python
from nltk.stem import RSLPStemmer

stemmer = RSLPStemmer()
stems = [stemmer.stem(t) for t in tokens_sem_stopwords_nltk]

for original, stem in zip(tokens_sem_stopwords_nltk, stems):
    print(f"{original:15} -> {stem}")
```

```python
doc_limpo = nlp(" ".join(tokens_sem_stopwords_nltk))

for token in doc_limpo:
    print(f"{token.text:15} -> {token.lemma_}")
```

```python
comparacao = pd.DataFrame({
    "Token original": tokens_sem_stopwords_nltk,
    "Stemming (RSLP)": stems,
    "Lematização (spaCy)": [t.lemma_ for t in doc_limpo]
})
comparacao
```
