```python
# Instalação e download de recursos (rodar 1x)
!pip install -q nltk spacy pandas
!python -m spacy download pt_core_news_sm -q

import nltk
import spacy
import pandas as pd

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('rslp', quiet=True)

nlp = spacy.load("pt_core_news_sm")

print("Ambiente configurado com sucesso.")
```

```python
frase_exercicio = "As reformas recentes valorizaram bastante o imóvel da esquina."

# 1. Tokenização com spaCy
doc_exercicio = nlp(frase_exercicio)
tokens = [token.text for token in doc_exercicio]
print("Tokens:", tokens)
```

```python
# 2. Remoção de pontuação de stopwords
from nltk.corpus import stopwords

stop_pt = set(stopwords.words('portuguese'))

tokens_limpos = [
    token.text for token in doc_exercicio
    if not token.is_punct and token.text.lower() not in stop_pt
]

print("Tokens após remover pontuação e stopwords:")
print(tokens_limpos)
```

```python
# 3. Stemming (RSLP)
from nltk.stem import RSLPStemmer

stemmer = RSLPStemmer()
stems = [stemmer.stem(token) for token in tokens_limpos]

for original, stem in zip(tokens_limpos, stems):
    print(f"{original} -> {stem}")
```

```python
# 4. Lematização (spaCy)
doc_limpo = nlp(" ".join(tokens_limpos))

lemmas = [token.lemma_ for token in doc_limpo]

for original, lemma in zip(tokens_limpos, lemmas):
    print(f"{original} -> {lemma}")
```

```python
# 5. Tabela comparativa
comparacao = pd.DataFrame({
    "Token original": tokens_limpos,
    "Stemming (RSLP)": stems,
    "Lematização (spaCy)": lemmas
})

comparacao
```
