```python
import spacy
import nltk
import pandas as pd
import numpy as np

from nltk.corpus import stopwords
from nltk.stem import RSLPStemmer

nltk.download('stopwords')
nltk.download('rslp')

nlp = spacy.load("pt_core_news_sm")
nlp_md = spacy.load("pt_core_news_md")
```

```python
corpus = [
    "Bom dia! Meu nome é Marcos Ferreira e tenho interesse no apartamento anunciado na Rua Voluntários da Pátria, em Matão.",
    "Vocês trabalham com financiamento pela Caixa Econômica Federal para imóveis usados?",
    "Gostaria de agendar uma visita ao imóvel do bairro Jardim Europa nesta semana.",
    "Qual é o valor do condomínio e do IPTU do apartamento no centro de Araraquara?",
    "A casa na Vila Mariana ainda está disponível para locação?",
    "Aceitam fiador ou seguro fiança para o contrato de aluguel?",
    "Sou a corretora Camila Souza, entrando em contato sobre o interesse do cliente João Pedro no imóvel de Ribeirão Preto.",
    "O apartamento é novo ou já foi reformado recentemente?",
    "Qual o prazo médio de aprovação do financiamento bancário?",
    "Estou buscando um imóvel próximo ao Shopping Iguatemi, para compra à vista.",
    "O sobrado no bairro Santa Cecília aceita animais de estimação?",
    "Poderiam enviar mais fotos do apartamento e da planta baixa?",
    "Qual é a taxa de juros aplicada pelo Banco do Brasil nesse tipo de financiamento?",
    "A casa alugada tem vaga de garagem coberta?",
    "Tenho interesse em investir em um imóvel para locação de longo prazo.",
    "Poderia me indicar imóveis disponíveis na região central de Matão?"
]
```

```python
# Processando todos os documentos do corpus
docs = [nlp(texto) for texto in corpus]
```

## 1. Tokenização

```python
# Etapa 1 — Tokenização

for i in [0, 1]:
    print(f"Documento {i+1}:")
    print([token.text for token in docs[i]])
    print()
```

## 2. Stopwords, Stemming e Lematização

```python
# ETAPA 2 - STOPWORDS, STEMMING E LEMATIZAÇÃO

# Escolhendo o Documento 1 do corpus
frase = corpus[0]
doc = nlp(frase)

# Stopwords em português
stop_pt = set(stopwords.words("portuguese"))

# Tokens sem pontuação e sem stopwords
tokens_filtrados = [
    token
    for token in doc
    if not token.is_punct
    and not token.is_space
    and token.text.lower() not in stop_pt
]

# Textos após remoção de stopwords
tokens_sem_stopwords = [
    token.text for token in tokens_filtrados
]

# Stemming aplicado somente aos tokens filtrados
stemmer = RSLPStemmer()

stems = [
    stemmer.stem(token.text.lower())
    for token in tokens_filtrados
]

# Lematização aplicada aos tokens filtrados
lemas = [
    token.lemma_.lower()
    for token in tokens_filtrados
]

print("Frase original:")
print(frase)

print("\nTokens sem pontuação e sem stopwords:")
print(tokens_sem_stopwords)

print("\nStemming:")
print(stems)

print("\nLematização:")
print(lemas)
```

```python
# Tabela comparativa

tabela_etapa2 = pd.DataFrame({
    "Original": tokens_sem_stopwords,
    "Sem Stopword": tokens_sem_stopwords,
    "Stemming": stems,
    "Lematização": lemas
})

tabela_etapa2
```

## 3. Análise Sintática

```python
# ETAPA 3 - ANÁLISE SINTÁTICA

frases_sintaticas = [
    corpus[1],
    corpus[3]
]

docs_sintaticos = [nlp(frase) for frase in frases_sintaticas]

for i, doc in enumerate(docs_sintaticos):
    print(f"\nFrase {i+1}:")
    print(doc.text)

    print("\nTokens, classe gramatical e dependência:")
    for token in doc:
        print(
            f"{token.text:20} "
            f"POS={token.pos_:10} "
            f"DEP={token.dep_:10} "
            f"HEAD={token.head.text}"
        )
```

```python
# Identificação da ROOT e análise do caso especial

for i, doc in enumerate(docs_sintaticos):

    print("\n" + "=" * 60)
    print(f"Frase {i+1}: {doc.text}")

    root = [token for token in doc if token.dep_ == "ROOT"][0]

    print("\nROOT:")
    print(f"Palavra: {root.text}")
    print(f"POS: {root.pos_}")
    print(f"Dependência: {root.dep_}")

    if root.pos_ == "VERB":
        print("A ROOT é um verbo comum.")

    elif any(token.dep_ == "cop" for token in doc):
        copula = [
            token.text
            for token in doc
            if token.dep_ == "cop"
        ]

        print("A ROOT não é um verbo comum.")
        print(
            f"É um caso especial de construção copular. "
            f"O verbo de cópula identificado foi: {', '.join(copula)}."
        )

    else:
        print(
            "A ROOT não é um verbo comum e deve ser interpretada "
            "considerando a estrutura específica da frase."
        )
```

```python
# Análise dos Sintagmas

for i, doc in enumerate(docs_sintaticos):

    print("\n" + "=" * 60)
    print(f"Frase {i+1}: {doc.text}")

    sujeitos = [
        token for token in doc
        if token.dep_ in ("nsubj", "nsubj:pass")
    ]

    roots = [
        token for token in doc
        if token.dep_ == "ROOT"
    ]

    print("\nSintagma Nominal (SN):")

    if sujeitos:
        for sujeito in sujeitos:
            sn = list(sujeito.subtree)
            print(" ".join(token.text for token in sn))
    else:
        print(
            "Não há sujeito nominal explícito identificado "
            "pelo modelo."
        )

    print("\nSintagma Verbal (SV):")

    if any(token.dep_ == "cop" for token in doc):
        print(
            "Construção copular: o verbo 'é' funciona como cópula "
            "e a estrutura não apresenta um SV convencional "
            "com verbo lexical como núcleo."
        )
    else:
        for root in roots:
            print(" ".join(token.text for token in root.subtree))
```

## 4. Interpretação Semântica

```python
# ETAPA 4 - INTERPRETAÇÃO SEMÂNTICA

pares = [
    ("apartamento", "imóvel"),
    ("financiamento", "aluguel")
]

for palavra1, palavra2 in pares:
    doc1 = nlp_md(palavra1)
    doc2 = nlp_md(palavra2)

    similaridade = doc1.similarity(doc2)

    print(f"{palavra1} x {palavra2}")
    print(f"Similaridade: {similaridade:.4f}")
    print()
```

```python
# Tabela de similaridade semântica

resultados_semanticos = []

for palavra1, palavra2 in pares:
    doc1 = nlp_md(palavra1)
    doc2 = nlp_md(palavra2)

    similaridade = doc1.similarity(doc2)

    resultados_semanticos.append({
        "Palavra 1": palavra1,
        "Palavra 2": palavra2,
        "Similaridade": round(similaridade, 4)
    })

tabela_semantica = pd.DataFrame(resultados_semanticos)

tabela_semantica
```

## 5. Extração de Features

```python
# ETAPA 5 - EXTRAÇÃO DE FEATURES

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
# Lematizando todos os documentos

corpus_lematizado = []

for doc in docs:

    lemas_doc = [
        token.lemma_.lower()
        for token in doc
        if not token.is_punct
        and not token.is_space
    ]

    corpus_lematizado.append(" ".join(lemas_doc))

print("Corpus lematizado:\n")

for i, texto in enumerate(corpus_lematizado):
    print(f"Documento {i+1}:")
    print(texto)
    print()
```

```python
# Removendo stopwords do corpus lematizado

corpus_lematizado_sem_stopwords = []

for texto in corpus_lematizado:

    palavras = texto.split()

    palavras_filtradas = [
        palavra
        for palavra in palavras
        if palavra.lower() not in stop_pt
    ]

    corpus_lematizado_sem_stopwords.append(
        " ".join(palavras_filtradas)
    )

print("Corpus lematizado sem stopwords:\n")

for i, texto in enumerate(corpus_lematizado_sem_stopwords):
    print(f"Documento {i+1}:")
    print(texto)
    print()
```

```python
# Bag-of-Words

bow_vectorizer = CountVectorizer()

bow_matrix = bow_vectorizer.fit_transform(
    corpus_lematizado_sem_stopwords
)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=bow_vectorizer.get_feature_names_out(),
    index=[
        f"Documento {i+1}"
        for i in range(len(corpus))
    ]
)

bow_df
```

```python
# Matriz TF-IDF

tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(
    corpus_lematizado_sem_stopwords
)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out(),
    index=[
        f"Documento {i+1}"
        for i in range(len(corpus))
    ]
)

tfidf_df
```

```python
# Quantidade de palavras únicas por documento

palavras_unicas = []

for i, texto in enumerate(corpus_lematizado_sem_stopwords):

    palavras = texto.split()
    unicas = set(palavras)

    palavras_unicas.append({
        "Documento": f"Documento {i+1}",
        "Palavras únicas": len(unicas)
    })

palavras_unicas_df = pd.DataFrame(palavras_unicas)

palavras_unicas_df
```

```python
# Documento com maior quantidade de palavras únicas

maior = palavras_unicas_df.loc[
    palavras_unicas_df["Palavras únicas"].idxmax()
]

print(
    f"O documento com mais palavras únicas é "
    f"{maior['Documento']}, com "
    f"{maior['Palavras únicas']} palavras únicas."
)
```

```python
# Encontrando o maior valor de TF-IDF

maior_valor = tfidf_df.to_numpy().max()

linha, coluna = np.unravel_index(
    tfidf_df.to_numpy().argmax(),
    tfidf_df.shape
)

documento_maior_tfidf = tfidf_df.index[linha]
palavra_maior_tfidf = tfidf_df.columns[coluna]

print("Documento:", documento_maior_tfidf)
print("Palavra:", palavra_maior_tfidf)
print("TF-IDF:", round(maior_valor, 4))
```

## 6. Descoberta de Conhecimento em Textos (KDT)

```python
# Documentos escolhidos para NER
documentos_ner = [
    corpus[0],
    corpus[6],
    corpus[12]
]

for i, texto in enumerate(documentos_ner):

    doc = nlp(texto)

    print("\n" + "=" * 70)
    print(f"Documento selecionado {i+1}")
    print(texto)

    print("\nEntidades encontradas:")

    if doc.ents:
        for ent in doc.ents:
            print(f"{ent.text} --> {ent.label_}")
    else:
        print("Nenhuma entidade encontrada.")
```

```python
# Palavras-chave de cada documento

for i in range(len(tfidf_df)):

    documento = tfidf_df.iloc[i]

    principais = documento.sort_values(
        ascending=False
    ).head(5)

    print("\n" + "=" * 60)
    print(f"Documento {i+1}")

    for palavra, valor in principais.items():
        print(
            f"{palavra}: {valor:.4f}"
        )
```

```python
from sklearn.decomposition import LatentDirichletAllocation
```

```python
# Treinamento do LDA

lda = LatentDirichletAllocation(
    n_components=2,
    random_state=42
)

lda_matrix = lda.fit_transform(bow_matrix)

print("LDA treinado com sucesso.")
```

```python
# Palavras principais de cada tópico

feature_names = bow_vectorizer.get_feature_names_out()

n_palavras = 8

for topico_idx, topico in enumerate(lda.components_):

    palavras = [
        feature_names[i]
        for i in topico.argsort()[
            :-n_palavras - 1:-1
        ]
    ]

    print(
        f"\nTópico {topico_idx + 1}:"
    )

    print(", ".join(palavras))
```

```python
# Distribuição dos tópicos por documento

distribuicao_lda = pd.DataFrame(
    lda_matrix,
    columns=[
        "Tópico 1",
        "Tópico 2"
    ],
    index=[
        f"Documento {i+1}"
        for i in range(len(corpus))
    ]
)

distribuicao_lda
```

```python
# Tópico predominante em cada documento

distribuicao_lda["Tópico predominante"] = (
    distribuicao_lda[
        ["Tópico 1", "Tópico 2"]
    ].idxmax(axis=1)
)

distribuicao_lda
```

## 7. Conclusão Integradora

As etapas do pipeline de PLN estão relacionadas e servem de base umas para as outras. A tokenização permitiu dividir os documentos em unidades que puderam ser analisadas individualmente. A normalização, por meio da remoção de stopwords, stemming e lematização, ajudou a reduzir variações das palavras e preparar o texto para a extração de características. A análise sintática acrescentou informações sobre a função das palavras e suas relações dentro das frases, enquanto a similaridade semântica permitiu comparar os significados de termos do corpus. Essas etapas foram importantes para as etapas finais, pois a lematização, por exemplo, foi utilizada antes da construção das matrizes Bag-of-Words e TF-IDF. Um exemplo concreto é a relação entre “apartamento” e “imóvel”: a análise semântica apresentou similaridade de 0,6667, enquanto o TF-IDF permitiu identificar quais palavras eram mais características de cada documento. Dessa forma, as técnicas estudadas ao longo do semestre formam um pipeline integrado para transformar textos em informações que podem ser analisadas e utilizadas na descoberta de conhecimento.
