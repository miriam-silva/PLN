```python
# Instalação e download de recursos (rodar 1x)
!pip install -q nltk spacy pandas
!python -m spacy download pt_core_news_md -q

import nltk
import spacy
import pandas as pd

nltk.download('stopwords', quiet=True)
nltk.download('rslp', quiet=True)

nlp = spacy.load("pt_core_news_md")
print("Ambiente configurado com sucesso.")
```

```python
frase_exercicio = "O proprietário alugou a casa para a família rapidamente."
```

```python
# Passo 1: Tokenização
doc_exercicio = nlp(frase_exercicio)
tokens = [token.text for token in doc_exercicio]

print("Tokens:")
print(tokens)
```

```python
# Passo 2: Remoção de pontuação e stopwords
tokens_limpos = [
    token.text
    for token in doc_exercicio
    if not token.is_punct and not token.is_stop
]

print("Tokens limpos:")
print(tokens_limpos)
```

```python
# Passo 3: Stemming e Lematização
from nltk.stem import RSLPStemmer

stemmer = RSLPStemmer()
stems = [stemmer.stem(token) for token in tokens_limpos]

doc_limpo = nlp(" ".join(tokens_limpos))
lemas = [token.lemma_ for token in doc_limpo]

comparacao = pd.DataFrame({
    "Token": tokens_limpos,
    "Stemming (RSLP)": stems,
    "Lematização (spaCy)": lemas
})

comparacao
```

```python
# Passo 4: POS Tagging
dados_pos = [
    {"Token": token.text, "POS": token.pos_}
    for token in doc_exercicio
]

pd.DataFrame(dados_pos)
```

```python
# Passo 5: Sintagma Nominal e Sintagma Verbal
raiz = [token for token in doc_exercicio if token.dep_ == "ROOT"][0]
sujeito = [token for token in doc_exercicio if token.dep_ == "nsubj"][0]

sintagma_nominal = list(sujeito.subtree)
sintagma_verbal = [
    token for token in raiz.subtree
    if token not in sintagma_nominal
]

print("Sintagma Nominal (SN):")
print([token.text for token in sintagma_nominal])

print("Sintagma Verbal (SV):")
print([token.text for token in sintagma_verbal])
```

```python
# Passo 6: Similaridade entre duas palavras da frase
palavra_a = nlp("proprietário")
palavra_b = nlp("família")

similaridade = palavra_a.similarity(palavra_b)

print(f"Similaridade proprietário x família: {similaridade:.3f}")
```
