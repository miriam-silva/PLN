```python
# Instalação e download de recursos (rodar 1x)
!pip install -q spacy nltk pandas
!python -m spacy download pt_core_news_sm -q

import nltk
import spacy
import pandas as pd

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

nlp = spacy.load("pt_core_news_sm")

print("Ambiente configurado com sucesso.")
```

```python
frase_exercicio = "Olá! Gostaria de agendar uma visita para ver o imóvel de 3 quartos amanhã às 14h?"

# 1. Tokenização com spaCy
doc_exercicio = nlp(frase_exercicio)

tokens_originais = [token.text for token in doc_exercicio]

print("Tokens originais:")
print(tokens_originais)
```

```python
# 2. Remoção de pontuação
tokens_sem_pontuacao = [token.text for token in doc_exercicio if not token.is_punct]

print("Tokens sem pontuação:")
print(tokens_sem_pontuacao)
```

```python
# 3. Contagem: original vs. sem pontuação
print("Contagem total de tokens originais:", len(tokens_originais))
print("Contagem de tokens sem pontuação:", len(tokens_sem_pontuacao))
```
